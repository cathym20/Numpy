## Numerical Data Analysis


Import Statement

```python
import numpy as np
```

```python
arr = np.random.randint(1,1000,size=(100,100))
arr
```

```python
print('Maximum of all rows', '\n', np.max(arr,axis=1),'\n',
      'Minimum of all rows', '\n', np.min(arr,axis=1),'\n',
      'Maximum of all columns', '\n', np.max(arr,axis=0),'\n',
      'Minimum of all columns', '\n', np.min(arr,axis=0))
```

```python
print(np.mean(arr,axis=0), np.std(arr,axis=0))
```

```python
cov = np.cov(arr,rowvar=False)
print('Covariance matrix shape:',cov.shape)
```

```python
corr = np.corrcoef(arr,rowvar=False)
print('Correlation matrix:', corr[1,:])
```
